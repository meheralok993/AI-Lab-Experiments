```python
from collections import deque

def bfs (graph, start) :

  visited = set()

  queue = deque([start])

  visited.add(start) # Corrected: use .add() method

  print("BFS traversal:", end=" ")

  while queue:
    vertex = queue.popleft()
    print(vertex, end=" ")

    # This loop should be inside the BFS function and process the current 'vertex'
    for neighbour in graph.get(vertex, []):
      if neighbour not in visited:
        visited.add(neighbour)
        queue.append(neighbour)


#main problem


graph = {} # Initialize graph at the correct scope

n = int(input("Enter the number of vertices :"))

for i in range(n):
  vertex = input(f"Enter vertex {i+1}:") # Corrected f-string for display
  neighbours = input(f"Enter adjacent vertices of {vertex} (space-separated):").split()
  graph[vertex] = neighbours

start = input("Enter the starting vertex:")

print("\nGraph:", graph)

bfs(graph, start)
```
