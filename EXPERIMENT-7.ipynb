```python
def solve_queens(row, board):
    if row == 8:
        return True

    for col in range(8):
        safe = True

        for r in range(row):
            if (board[r] == col or
                abs(board[r] - col) == abs(r - row)):
                safe = False
                break

        if safe:
            board[row] = col

            if solve_queens(row + 1, board):
                return True

    return False

board = [-1] * 8
solve_queens(0, board)

print("One solution:")
for row in range(8):
    line = ['.'] * 8
    line[board[row]] = 'Q'
    print(' '.join(line))
```
