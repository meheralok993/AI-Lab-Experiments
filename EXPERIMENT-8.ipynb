```python
from collections import deque

def water_jug(cap1, cap2, target):
    queue = deque([(0, 0)])
    parent = {(0, 0): None}

    while queue:
        a, b = queue.popleft()

        if a == target or b == target:
            path = []
            state = (a, b)

            while state is not None:
                path.append(state)
                state = parent[state]

            return path[::-1]

        states = [
            (cap1, b),
            (a, cap2),
            (0, b),
            (a, 0),
            (a - min(a, cap2 - b), b + min(a, cap2 - b)),
            (a + min(b, cap1 - a), b - min(b, cap1 - a))
        ]

        for state in states:
            if state not in parent:
                parent[state] = (a, b)
                queue.append(state)

    return None

path = water_jug(4, 3, 2)

if path:
    for state in path:
        print(state)
else:
    print("Solution not possible")
```
